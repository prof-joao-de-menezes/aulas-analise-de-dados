# Lidando com o Numpy
O numpy possibilita criar arrays, que são listas onde podemo fazer lógicas e operações dentro de cada valor da lista, mesmo sem a necessidade de especificar o index.

Isso significa que arrays são focados em calculos matemáticos e manipulação de um grande volume de dados.

```python
import numpy as np
#                jan    fev    mar   abr
renda = np.array([5000, 4500, 6000, 4000])
gastos = np.array([4000,3500, 6000, 3900])

lucro = renda - gastos

print(f" Lucro dos meses: {lucro}")
```

```python
notas_alunos = np.array([
    10.0,6.0,5.6, 7.4, 8.8
])

notas_aprovados = notas_alunos[notas_alunos >= 7]
print(f"Notas aprovadas: {notas_aprovados}")

notas_reprovados = notas_alunos[~(notas_alunos >= 7)]
print(f"Notas reprovadas: {notas_reprovados}")
```

```python
vendas = np.array([
    [15000, 20000, 30000], # 1° trimestre
    [25000, 30000, 40000], # 2° trimestre
    [35000, 40000, 50000], # 3° trimestre
    [45000, 50000, 60000], # 4° trimestre
])

total_vendas_ano = np.sum(vendas)
print(total_vendas_ano)

# axis=1 -> faz o calculo da função por linha
total_por_trimestre = np.sum(vendas, axis=1)
print(total_por_trimestre)

comeco_de_trimestre = np.sum(vendas, axis=0)
print(comeco_de_trimestre)

maior_venda_ano = np.max(vendas, axis=1)
print(f"Maior venda ano: {maior_venda_ano}")
menor_venda_ano = np.min(vendas)
print(f'Menor venda ano: {menor_venda_ano}')
```

```python
valor_produtos = np.array([
    10,15,20,30
])

#       np.where(condição, True, False)
valor_promocao = np.where(
    valor_produtos >= 20, valor_produtos * 0.9, valor_produtos
)
valor_promocao = np.round(valor_promocao, 2)
print(valor_promocao)
```

```python
import pandas as pd

df_produtos = pd.DataFrame(
    {
        "Produto": ['Notebook', 'Celular', 'Monitor', "Notebook"],
        "Estoque": [10, 0, 5, 9],
        "Valor": [2000, 1500, 500, 5000],
        "Cliente": ["Magalu", "Americanas", "Kabum", "Magalu"]
    }
)

df_produtos["Valor"] = np.where(
    df_produtos["Valor"] > 1500,
    df_produtos["Valor"] * 0.9,
    df_produtos["Valor"]
)
df_produtos["Valor"] = np.where(
    df_produtos["Cliente"] == "Kabum",
    df_produtos["Valor"] * 0.8,
    df_produtos["Valor"]
)

df_produtos["Estoque"] = np.where(
    df_produtos["Cliente"] == "Kabum",
    df_produtos["Estoque"] * 0,
    np.where(
        df_produtos["Cliente"] == "Magalu",
        df_produtos["Estoque"] - 2,
        df_produtos["Estoque"]
    )
)

df_produtos["Status"] = np.where(
    df_produtos["Estoque"] > 0,
    "Disponível",
    "Esgotado"
)

print(df_produtos)
```

```python
agrupamento_count = df_produtos.groupby(["Cliente", "Estoque"])["Valor"].sum()

print(agrupamento_count)
```
