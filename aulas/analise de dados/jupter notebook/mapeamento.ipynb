```python
import pandas as pd
import numpy as np
# criando uma base de dados
dados_alunos = [
    {"nome": "João", "turno": "N", "nota": 8.5},
    {"nome": "Marcelo", "turno": "T", "nota": 7.5},
    {"nome": "Victor", "turno": "N", "nota": 5.5},
    {"nome": "Bruno", "turno": "M", "nota": 9.5},
    {"nome": "Rafael", "turno": "M", "nota": 10.0}
]
# criando um dataframe
df_alunos = pd.DataFrame(dados_alunos)
print(df_alunos)
```

```python
# inserir coluna por mapeamento
map_turnos = {
    "M": "Manhã",
    "T": "Tarde",
    "N": "Noite"
}

df_alunos["turno_completo"] = df_alunos["turno"].map(map_turnos)
print(df_alunos)
```

```python
# numpy -> np
# np.where(condição, verdadeiro, false)
df_alunos["nota"] = np.where(
    df_alunos["nota"] >=5,"Aprovado","Reprovado"
)
print(df_alunos)
```

```python
df_alunos["troca_turno"] = np.where(
    df_alunos["status"] == "Aprovado",
    "Concluíu o esino médio",
    "Tem que fazer a prova de recuperação"
)

print(df_alunos)
```

```python
#escrever um arquivo csv
df_alunos.to_csv("alunos_data.csv", index=False)
print(df_alunos)
```
