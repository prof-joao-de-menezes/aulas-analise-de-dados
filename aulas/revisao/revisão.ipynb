# enumerating e range


```python
produtos = [
    "Telefone",
    "Notebook",
    "Televisão",
    "Notebook",
    "Caixa de som",
    "Televisão"
]

# separando com o range e o len
print("FOR COM RANGE")
for index in range(len(produtos)):
    print(f"Index do produto: {index} |"
          f" Produto {produtos[index]}")

# enumerate separa indices e valores
print("FOR COM ENUMERATE")
for indice, valor in enumerate(produtos, start=1):
    print(f'Indice {indice} | Valor {valor}')

dicionario = {
    "nome": "Joao",
    "idade": 22,
    "endereco": "Luziania",
}

contador = 0
while contador < len(produtos):
    print(f'Indice {contador} | Valor {produtos[contador]}')
    contador += 1
```

# Type hinting

diferente do casting, que muda o tipo da variável, o type hinting especifica como um comentário qual tipo de dado a função quer receber como parâmetro.

```python
# SEM O TYPE HINTING
def calculo_media(notas):
    return sum(notas) / len(notas)

# COM O TYPE HINTING
def calculo_media_tipada(notas: list[float]) -> float:
    return sum(notas) / len(notas)

notas = [10.0,8.5,9.6,6.8]

print(calculo_media_tipada(notas))
```

# List Comprehension (Compressão de lista)
Não muda nada na execução, apenas diminui o código do for, e retira a necessidade de usar o .append

```python
# SEM COMPRESSÃO
temperaturas = [-10, 18, 30, -90, 50]
temperaturas_validas = []
for temperatura in temperaturas:
    if temperatura > 0:
        fahrenheit = (temperatura * 1.0) + 32
        temperaturas_validas.append(fahrenheit)
print(temperaturas_validas)
```

```python
# COM COMPRESSÃO
temperaturas = [-10, 18, 30, -90, 50]
temperaturas_validas = [
    (temperatura * 1.0) + 32
    for temperatura in temperaturas
    if temperatura > 0
]
print(temperaturas_validas)
```

# DECORATOR (Deoradores)

```python
# criando decorador
def meu_decorador(funcao):
    def executar():
        print("="*20)
        print("Iniciando aplicação...")
        funcao()
        print("Finalizando aplicação...")
        print("="*20)
    return executar

# APLICANDO DECORADOR
@meu_decorador
def olaMundo():
    print("Olá mundo")

olaMundo()
```
